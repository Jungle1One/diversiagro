# 🌳 Análisis y Validación de Cobertura de la Tierra (IDEAM 2022)

**Objetivo**: Cargar el Shapefile de Coberturas de la Tierra a escala 1:25.000, validarlo cruzándolo con los Conglomerados de Vocación (MMRA) del DANE y explorar su utilidad como máscara de exclusión y variable predictiva en nuestro modelo de Machine Learning.

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['figure.dpi'] = 100
sns.set_style('whitegrid')

BASE_DIR = Path('.').resolve().parent
PROCESSED_DIR = BASE_DIR / 'data' / 'processed'
REPORTS_DIR = BASE_DIR / 'data' / 'reports'

# Archivos de entrada
MMRA_FILE = BASE_DIR / 'data' / 'raw' / 'mmra_conglomerados_valle.geojson'
COBERTURA_FILE = PROCESSED_DIR / 'Coberturas_de_la_Tierra_2022_SI_Escala_1_25000' / 'Coberturas_de_la_Tierra_25K_2022_SI.shp'

print('Configuración lista.')
```

## 1. Carga de la Máscara Base (MMRA Palmira)
El mapa de coberturas de la tierra pesa más de 300MB porque contiene un altísimo nivel de detalle. Para no colapsar la memoria RAM, usaremos una técnica de filtrado espacial nativo (`mask`) para cargar **únicamente** los polígonos de cobertura que caigan sobre el municipio de Palmira.

```python
# 1.1 Cargar MMRA completo y filtrar Palmira
print("Cargando MMRA...")
gdf_mmra = gpd.read_file(MMRA_FILE)
gdf_mmra['NOMBREMUNICIPIO_CG'] = gdf_mmra['NOMBREMUNICIPIO_CG'].astype(str).str.strip().str.upper()

municipio_objetivo = 'PALMIRA'
gdf_mmra_local = gdf_mmra[gdf_mmra['NOMBREMUNICIPIO_CG'] == municipio_objetivo]

print(f"Polígonos MMRA en {municipio_objetivo}: {len(gdf_mmra_local)}")
```

## 2. Extracción Espacial del Shapefile (Corine Land Cover)
Le decimos a GeoPandas que lea el disco duro pero que solo extraiga los pixeles/polígonos que toquen nuestra `mask` (los bordes de Palmira).

```python
print(f"Extrayendo coberturas de la tierra (IDEAM) espacialmente para {municipio_objetivo}...")
# Le pasamos el gdf_mmra_local como 'mask' para que filtre al momento de leer
gdf_cob = gpd.read_file(COBERTURA_FILE, mask=gdf_mmra_local)

# Validar y ajustar Sistema de Coordenadas (CRS) si son distintos
if gdf_cob.crs != gdf_mmra_local.crs:
    print(f"Reproyectando coberturas de {gdf_cob.crs} a {gdf_mmra_local.crs}")
    gdf_cob = gdf_cob.to_crs(gdf_mmra_local.crs)

print(f"¡Extraído! {len(gdf_cob)} polígonos de cobertura detallada encontrados.")
display(gdf_cob[['cobertura', 'cob_agrup', 'area_ha']].sample(5))
```

## 3. Visualización Cartográfica
Observemos cómo se ve la realidad física de la tierra (IDEAM) frente a los bloques administrativos del MMRA (DANE).

```python
fig, ax = plt.subplots(figsize=(16, 12))

# 1. Pintar las coberturas reales del suelo
gdf_cob.plot(column='cob_agrup',
             ax=ax,
             legend=True,
             cmap='Set2',
             edgecolor='none',
             alpha=0.8,
             legend_kwds={'bbox_to_anchor': (1.05, 1), 'loc': 'upper left', 'title': 'Grupo de Cobertura'})

# 2. Sobreponer los bordes del MMRA (Vocación agrícola)
gdf_mmra_local.plot(ax=ax, facecolor='none', edgecolor='black', linewidth=1.5, linestyle='--')

ax.set_title(f'Realidad (Cobertura 2022) vs Vocación (Bordes MMRA) - {municipio_objetivo}', fontsize=18, fontweight='bold')
ax.axis('off')
plt.tight_layout()
plt.savefig(REPORTS_DIR / 'fig_mapa_cobertura_vs_mmra.png', dpi=150)
plt.show()
```

## 4. Validación: Cruce Físico (Ground Truth)
Vamos a cruzar ambos mapas. Queremos ver, de todo lo que el DANE dice que tiene vocación de "Cultivos Transitorios" o "Pastos" en el MMRA, **qué hay realmente ahí hoy** según el satélite del IDEAM.

```python
print("Calculando intersección espacial profunda (Overlay)...")
interseccion = gpd.overlay(gdf_mmra_local, gdf_cob, how='intersection')

# Recalcular el área real en Hectáreas de los recortes usando proyección oficial métrica de Colombia
interseccion['area_real_ha'] = interseccion.to_crs("EPSG:3116").geometry.area / 10000

print("\n--- VALIDACIÓN: Vocación (DANE) vs Realidad (IDEAM) ---")
cruce = interseccion.groupby(['USO_PRED_CONGL', 'cob_agrup'])['area_real_ha'].sum().unstack().fillna(0)
cruce['Total_Vocacion_Ha'] = cruce.sum(axis=1)
cruce = cruce.sort_values('Total_Vocacion_Ha', ascending=False)

# Mostrar con mapa de calor interno para destacar las coincidencias
display(cruce.style.background_gradient(cmap='Greens', axis=1).format('{:,.1f}'))
```

### 🔎 Conclusión del Análisis:
1. **Discrepancias (Si las hay)**: Si ves que una zona de "Cultivos Permanentes" del MMRA tiene hectáreas de "Tejido Urbano Continuo", significa que la ciudad se ha comido la frontera agrícola y tu modelo de Machine Learning debe penalizar esa zona.
2. **Máscaras para el Modelo**: Gracias a este cruce, el modelo sabrá que no debe recomendar sembrar Cacao en zonas marcadas bajo la cobertura `Bosque denso`. El modelo aprenderá que `cob_agrup` = 'Territorios artificializados' equivale a $Y=0$ (Inviable).
